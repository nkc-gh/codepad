# Linear Regression Example

Model: `ŷ = wx + b`
- `w` = weight (slope)
- `b` = bias (intercept)
- `x` = input feature (house size)
- `y` = target output (house price)

```python
# Imports
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Training Data
# x_train = house sizes (1000 sqft)
# y_train = house prices ($1000s)
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])

# m = number of training examples
print(f"x_train.shape: {x_train.shape}") # returns a tuple
m = x_train.shape[0]

print(f"Number of training examples: m = {m}")
print(f"x_train: {x_train}")
print(f"y_train: {y_train}")
```

```python
# Access individual training examples
# x^(i) in ML = i-th training example (NOT power)
i = 0   # change to 1 to see second example
x_i = x_train[i]
y_i = y_train[i]
print(f"Example {i}: (x^({i}), y^({i})) = ({x_i}, {y_i})")
```

```python
# Model Function
# Computes ŷ = wx + b for all training examples
def compute_model_output(x, w, b):
    m = x.shape[0]          # number of training examples
    f_wb = np.zeros(m)      # empty array to store predictions

    for i in range(m):
        f_wb[i] = w * x[i] + b     # apply ŷ = wx + b

    return f_wb
```

```python
# Set model parameters and predict
# w=200, b=100 gives perfect predictions for this data
# House 1: ŷ = 200 * 1.0 + 100 = 300 ✅
# House 2: ŷ = 200 * 2.0 + 100 = 500 ✅
w = 200     # slope (weight)
b = 100     # intercept (bias)

predictions = compute_model_output(x_train, w, b)

print(f"w = {w}, b = {b}")
print(f"Predictions : {predictions}")
print(f"Actual      : {y_train}")
```

```python
# Plot — Data Points + Prediction Line
# scatter → plots raw data points (red X)
# plot    → draws the model's prediction line
plt.scatter(x_train, y_train, marker='x', c='r', label='Actual prices')
plt.plot(x_train, predictions, label='Prediction line (ŷ = wx + b)')

plt.title("Housing Prices — Linear Regression")
plt.xlabel("Size (1000 sqft)")
plt.ylabel("Price (in $1000s)")
plt.legend()
plt.show()
```
